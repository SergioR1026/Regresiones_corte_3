```python
# prompt: carga este excel/content/drive/MyDrive/datos_industrias.xlsx

import pandas as pd
df = pd.read_excel('/content/drive/MyDrive/datos_industrias.xlsx')
print(df.head())
```

```python
# prompt: necesito un analisis de regresion lineal simple que la variable X sea Salario y la variable Y sea Tasa_Crecimiento

import statsmodels.api as sm
import matplotlib.pyplot as plt

# Define dependent (Y) and independent (X) variables
X = df['Salario']
y = df['Tasa_Crecimiento']

# Add a constant to the independent variable for the intercept
X = sm.add_constant(X)

# Create and fit the OLS (Ordinary Least Squares) model
model = sm.OLS(y, X).fit()

# Print the regression summary
print(model.summary())

# Plot the regression line
plt.scatter(X['Salario'], y)
plt.plot(X['Salario'], model.predict(X), color='red')
plt.xlabel('Salario')
plt.ylabel('Tasa_Crecimiento')
plt.title('Regresión Lineal Simple: Tasa de Crecimiento vs Salario')
plt.show()
```
